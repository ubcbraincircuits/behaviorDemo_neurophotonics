```python
import cv2, os
video = r"D:\pymouse_lifter\PyMouse-Lifter\video\short_example.mp4"
outdir = r"D:\pymouse_lifter\PyMouse-Lifter\short_example_frames"
os.makedirs(outdir, exist_ok=True)

cap = cv2.VideoCapture(video)
i = 1
while True:
    ok, frame = cap.read()
    if not ok:
        break
    cv2.imwrite(os.path.join(outdir, f"{i:012d}.png"), frame)
    i += 1
cap.release()
print("saved", i, "frames")
```
